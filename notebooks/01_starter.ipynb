```python
import pandas as pd
from pathlib import Path
import glob as gb
import numpy as np
```

```python
data_dir = gb.glob("../data/raw/*.csv")

datasets = {}
for file_path in data_dir:
    datasets[Path(file_path).name] = pd.read_csv(file_path)

print(f"Loaded {len(datasets)} files")
    
```

```python
for key,value in datasets.items():
    print("Data set file", key)
```

```python
for key,value in datasets.items():
    df = value
    print("Data set file", key)
    print(df[' Label'].value_counts())
    print(df[' Label'].value_counts(normalize=True))
    print("Unique Label Type",df[' Label'].unique())
    print("====================================================================")
```

```python
for key,value in datasets.items():
    df = value
    print("Data set file", key)
    for col in df.columns:
        if df[col].isnull().sum() > 0:
            print(f"Column '{col}' has {df[col].isnull().sum()} missing values.")
        numeric_col = pd.to_numeric(df[col], errors='coerce') # Convert the column to numeric, coercing errors to NaN (need more explanation)
        if np.isposinf(numeric_col).sum() > 0:
            print(f"Column '{col}' contains {np.isposinf(numeric_col).sum()} positive infinite values.")
        if np.isneginf(numeric_col).sum() > 0:
            print(f"Column '{col}' contains negative infinite values.")
    print("====================================================================")
```

```python
for key,value in datasets.items():
    df = value
    print("Data set file", key)
    print(f"Number of duplicate rows: {df.duplicated().sum()}")
    print("====================================================================")
```

```python
for key,value in datasets.items():
    df = value
    print("Data set file", key)
    df.columns = df.columns.str.strip()
    df.columns = df.columns.str.replace(' ', '_')
    df['Label'] = df['Label'].str.replace('�', '_')
    df['Label'] = df['Label'].str.replace(' ', '_')
    print(df.columns)
    print(df['Label'].unique())
    print("====================================================================")
```

```python
for key,value in datasets.items():
    df = value
    df.drop_duplicates(inplace=True)
    df.replace(np.inf, np.nan, inplace=True)
    numeric_cola = df.select_dtypes(include=np.number).columns # explain this line
    df[numeric_cola] = df[numeric_cola].fillna(df[numeric_cola].median()) # explain this line
    print("Data set file", key)
    print(f"Number of duplicate rows after dropping duplicates: {df.duplicated().sum()}")
    print(f"Number of missing values after filling: {df.isnull().sum().sum()}")
    print(f"Number of infinite values after replacing: {np.isinf(df.select_dtypes(include=[np.number])).sum().sum()}") # what df.select_dtypes(include=[np.number]) returns
    print("====================================================================")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix 
from sklearn.metrics import f1_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import classification_report
from imblearn.over_sampling import SMOTE
from imblearn.over_sampling import BorderlineSMOTE
```

```python
x_temp , x_test , y_temp, y_test , labels_temp, labels_test = train_test_split(x, y, labels, test_size=0.2, random_state=42, stratify=y)
print(f"Training set shape: {x_temp.shape}, {y_temp.shape}")
print(f"Test set shape: {x_test.shape}, {y_test.shape}")
```

```python
x_train , x_val , y_train, y_val , labels_train, labels_val = train_test_split(x_temp, y_temp, labels_temp, test_size=0.125, random_state=42, stratify=y_temp)
print(f"Training set shape: {x_train.shape}, {y_train.shape}")
print(f"Validation set shape: {x_val.shape}, {y_val.shape}")
print(f"Test set shape: {x_test.shape}, {y_test.shape}")
```

```python
print((y_train.value_counts().values[0]/len(y_train))*100)
print((y_train.value_counts().values[1]/len(y_train))*100)
```

```python
benign_mask = (y_train == 0)
malicious_mask = (y_train == 1)

x_train_benign = x_train[benign_mask]
x_train_malicious = x_train[malicious_mask]

y_train_benign = y_train[benign_mask]
y_train_malicious = y_train[malicious_mask]
print(f"Benign samples: {len(x_train_benign)}, Malicious samples: {len(x_train_malicious)}")
range_samples = x_train_benign.shape[0] // 2
print(f"Number of samples to keep from benign class: {range_samples}")
np.random.seed(42)
x_benign_sampled = np.random.choice(x_train_benign.index, size=range_samples, replace=False)
x_benign_undersampled = x_train_benign.loc[x_benign_sampled]
y_benign_undersampled = y_train_benign.loc[x_benign_sampled]

x_train_balanced = pd.concat([x_benign_undersampled, x_train_malicious], axis=0)
y_train_balanced = pd.concat([y_benign_undersampled, y_train_malicious], axis=0)
print(f"Balanced training set shape: {x_train_balanced.shape}, {y_train_balanced.shape}")
```

```python
sm = SMOTE(random_state=42)
x_train_smote, y_train_smote = sm.fit_resample(x_train_balanced,y_train_balanced)
print(f"SMOTE balanced training set shape: {x_train_smote.shape}, {y_train_smote.shape}")
```

```python
print("Original class distribution in training set:")
print(y_train.value_counts())
print("\nClass distribution after undersampling:")
print(y_train_balanced.value_counts())
print("\nClass distribution after SMOTE oversampling:")
print(y_train_smote.value_counts())
```

```python
print("Validation set class counts:\n", y_val.value_counts())
print("\nTest set class counts:\n", y_test.value_counts())
print("\nValidation shape:", x_val.shape)
print("Test shape:", x_test.shape)
```

```python
model = RandomForestClassifier(class_weight='balanced',random_state=42, n_estimators=100)
model.fit(x_train_balanced, y_train_balanced)
```

```python
preds = model.predict(x_val)
accuracy = accuracy_score(y_val, preds)
print(f"Validation Accuracy: {accuracy:.4f}")
f1 = f1_score(y_val, preds)
print(f"Validation F1 Score: {f1:.4f}")
conf_matr = confusion_matrix(y_val , preds)
print("Confusion Matrix:", conf_matr)
per_score = precision_score(y_val, preds)
print(f"Validation Precision Score: {per_score:.4f}")
rec_score = recall_score(y_val, preds)
print(f"Validation Recall Score: {rec_score:.4f}")
class_report = classification_report(y_val, preds)
print("Classification Report:\n", class_report)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(x_train_smote)
x_train_scaled = pd.DataFrame(scaler.transform(x_train_smote), columns=x_train_smote.columns)
x_val_scaled = pd.DataFrame(scaler.transform(x_val), columns=x_val.columns)
x_test_scaled = pd.DataFrame(scaler.transform(x_test), columns=x_test.columns)

print("Scaled train shape:", x_train_scaled.shape)
print("Scaled val shape:", x_val_scaled.shape)
print("Scaled test shape:", x_test_scaled.shape)
print("Scaled train data:")
print(x_train_scaled.head())
```
